# 🌸 IRIS FLOWER PREDICTION - COMPLETE PIPELINE

# 📦 STEP 1: Import Required Libraries

```python
import pandas as pd              # For data manipulation
import numpy as np               # For numerical operations
import matplotlib.pyplot as plt  # For data visualization
import joblib                    # For saving/loading ML models

from sklearn.datasets import load_iris                          # Iris dataset
from sklearn.model_selection import train_test_split            # Split data
from sklearn.preprocessing import StandardScaler                # Feature scaling
from sklearn.ensemble import RandomForestClassifier             # ML model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

# 📊 STEP 2: Load the Iris Dataset

```python
iris = load_iris()
# Built-in dataset with 150 samples, 4 features, 3 classes
```

```python
# Convert to a pandas DataFrame for easier handling
df = pd.DataFrame(
    iris.data,                    # Feature values
    columns=iris.feature_names    # Column names (sepal/petal length/width)
)
```

```python
# Add the target (species) column
df["species"] = iris.target
```

```python
# Preview first 5 rows
print("🔹 First 5 rows of the dataset:")
df.head()
```

# 🔍 STEP 3: Exploratory Data Analysis (EDA)

```python
# --- 3.1: Check dataset shape (rows, columns) ---
print(f"📐 Dataset shape: {df.shape}")
```

```python
# --- 3.2: Check data types and missing values ---
print("\n📋 Dataset info:")
df.info()
```

```python
# --- 3.3: Check for missing values ---
print("\n❓ Missing values per column:")
print(df.isnull().sum())
```

```python
# --- 3.4: Statistical summary of the data ---
print("\n📊 Statistical summary:")
df.describe()
```

# 📈 STEP 4: Data Visualization

```python
plt.figure(figsize=(8, 5))

# Scatter plot: Sepal Length vs Petal Length colored by species
plt.scatter(
    df["sepal length (cm)"],
    df["petal length (cm)"],
    c=df["species"],       # Color by species (0, 1, 2)
    cmap="viridis",
    edgecolors="black"
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.title("Iris Dataset - Sepal vs Petal Length")
plt.colorbar(label="Species")
plt.grid(alpha=0.3)
plt.show()
```

# 🧹 STEP 5: Data Cleaning

```python
# Remove duplicate rows (if any)
df = df.drop_duplicates().copy()   # .copy() avoids SettingWithCopyWarning
```

```python
print(f"✅ Shape after removing duplicates: {df.shape}")
```

# ⚙️ STEP 6: Feature Engineering

```python
# Create new features: area = length × width (captures combined size info)
df["sepal_area"] = df["sepal length (cm)"] * df["sepal width (cm)"]
df["petal_area"] = df["petal length (cm)"] * df["petal width (cm)"]
```

```python
# Preview engineered features
print("\n🔹 Dataset with new engineered features:")
df.head()
```

# 🎯 STEP 7: Separate Features (X) and Target (y)

```python
X = df.drop("species", axis=1)   # Features: 6 columns (4 original + 2 engineered)
y = df["species"]                # Target: species label (0, 1, 2)
```

```python
print(f"✅ Features shape: {X.shape}")
print(f"✅ Target shape: {y.shape}")
print(f"✅ Feature names: {list(X.columns)}")
```

# ⚖️ STEP 8: Feature Scaling
1.  StandardScaler: transforms features to have mean=0 and std=1
2. This improves model performance (especially for distance-based algorithms)

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"✅ Scaled features shape: {X_scaled.shape}")
```

# ✂️ STEP 9: Train-Test Split

1.  80% for training, 20% for testing
2. random_state=42 ensures reproducibility

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print(f"✅ Training set: {X_train.shape}")
print(f"✅ Testing set:  {X_test.shape}")
```

# 🤖 STEP 10: Model Training
1.  RandomForestClassifier: ensemble of 100 decision trees
2. Works well for classification, handles non-linear patterns

```python
model = RandomForestClassifier(
    n_estimators=100,     # Number of trees
    random_state=42       # Reproducibility
)
```

```python
# Train the model on training data
model.fit(X_train, y_train)

print("✅ Model trained successfully!")
```

# 📏 STEP 11: Model Evaluation

```python
# Make predictions on the test set
predictions = model.predict(X_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, predictions)
print(f"\n🎯 Accuracy: {accuracy * 100:.2f}%")
```

```python
# Detailed classification report (precision, recall, F1-score)
print("\n📋 Classification Report:")
print(classification_report(
    y_test,
    predictions,
    target_names=iris.target_names
))
```

```python
# Confusion matrix
print("\n🔢 Confusion Matrix:")
print(confusion_matrix(y_test, predictions))
```

```python
# Feature importance (which features matter most?)
print("\n⭐ Feature Importances:")
for name, importance in zip(X.columns, model.feature_importances_):
    print(f"   {name}: {importance:.4f}")
```

# 💾 STEP 12: Save Model and Scaler
==========================================
1. joblib is preferred over pickle for scikit-learn models (more efficient)

```python
joblib.dump(model, "model.pkl")
joblib.dump(scaler, "scaler.pkl")

print("✅ Model saved as 'model.pkl'")
print("✅ Scaler saved as 'scaler.pkl'")
```

# 🧪 STEP 13: Test the Saved Model (Load & Predict)

```python
# Load the saved model and scaler
loaded_model = joblib.load("model.pkl")
loaded_scaler = joblib.load("scaler.pkl")
```

```python
# Sample input: [sepal_length, sepal_width, petal_length, petal_width,
#                 sepal_area, petal_area]
sample = np.array([[5.1, 3.5, 1.4, 0.2, 5.1*3.5, 1.4*0.2]])
```

```python
# Scale the input using the loaded scaler
sample_scaled = loaded_scaler.transform(sample)
```

```python
# Predict
prediction = loaded_model.predict(sample_scaled)[0]
predicted_species = iris.target_names[prediction]
```

```python
print(f"\n🌸 Sample input: {sample[0]}")
print(f"🎯 Predicted species: {predicted_species}")
```
